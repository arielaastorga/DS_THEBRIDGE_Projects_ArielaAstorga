```python
#EJEMPLO GET

import requests

root_path = "http://127.0.0.1:5000"
path = "/predict_query"

params = {
    "neighbourhood": "Sol",
    "room_type": "Entire home/apt",
    "minimum_nights": 5,
    "number_of_reviews": 173,
    "availability_365": 342,
    "number_of_reviews_ltm": 1
}

response = requests.get(url=root_path + path, params=params)

print(response)
print(response.text)
print(response.json())
```

```python
#EJEMPLO POST

import requests

root_path = "http://127.0.0.1:5000"
path = "/predict"

content = {
    "neighbourhood": "Sol",
    "room_type": "Entire home/apt",
    "minimum_nights": 5,
    "number_of_reviews": 173,
    "availability_365": 342,
    "number_of_reviews_ltm": 1
}

response = requests.post(url=root_path + path, json=content)

print(response)
print(response.text)
print(response.json())
```
