```python
import requests
from pprint import pprint

url = "https://online-ds-bridge-proyects-arielaastorga.onrender.com/predict-query"

params = {
    "accommodates": 6,
    "bathrooms": 2,
    "bedrooms": 3,
    "beds": 4,
    "minimum_nights": 1,
    "number_of_reviews": 80,
    "review_scores_rating": 5,
    "availability_365": 250,
    "neighbourhood_cleansed": "Justicia",
    "room_type": "Entire home/apt"
}

r = requests.get(url, params=params)

print("Status code:", r.status_code)
print("URL final:", r.url)
pprint(r.json())
```

```python
# EJEMPLO POST

import requests

url = "https://online-ds-bridge-proyects-arielaastorga.onrender.com/predict"

payload = {
    "accommodates": 4,
    "bathrooms": 1,
    "bedrooms": 2,
    "beds": 2,
    "minimum_nights": 2,
    "number_of_reviews": 50,
    "review_scores_rating": 4,
    "availability_365": 180,
    "neighbourhood_cleansed": "Sol",
    "room_type": "Entire home/apt"
}

response = requests.post(url, json=payload)

print("Status code:", response.status_code)
print(response.json())
```
